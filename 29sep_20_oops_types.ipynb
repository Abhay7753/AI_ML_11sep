```python
class Rectangle:
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def area(self):
    return self.length*self.width
r=Rectangle(5,6)

print(r.area())
```

```python
class Bank:

  def __init__(self,balance):
    self.balance=balance
  def deposit(self,amount):
    self.balance+=amount
account=Bank(1000)
account.deposit(500)
print(account.balance)
```

```python
class Student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def result(self):
    if self.marks>=40:
      print("Pass")
    else:
      print("Fail")


s2=Student("Rahul",25)

s2.result()
```

```python
class Student:

  def __init__(self,name):
    self.name=name

  def change_name(self,new_name):
    self.name=new_name
s=Student("Rahul")

print(s.name)

s.change_name("Amit")
print(s.name)
```

```python
class Counter:
  def __init__(self):
    self.count=0
  def increase(self):
    print(self.count)
    self.count+=1
c=Counter()

for i in range(5):
  c.increase()
print(c.count)
```

# Inheritance

```python
# Parent class ---> THe class we are inheriting from ---> base class

# chil class --->  the class that inherits from another class ---> derived class

```

```python
class Person:
  def __init__(self,first_name,last_name):
    self.first_name=first_name
    self.last_name=last_name
  def show_name(self):
    print(self.first_name,self.last_name)
```

```python
p1=Person("Raj","Sharma")

p1.show_name()
```

```python
class Student(Person):
  pass
```

```python
s1=Student("Abhya jeet ","Chauhan")
```

```python

s1.show_name()
```

```python
class Vehicles:
  def __init__(self,brand):
    self.brand=brand
  def show(self):
    print(self.brand)
```

```python
s1=Vehicles("SWift")
s1.show()
```

```python
class Car(Vehicles):
  car_engine="xyzz"
```

```python
s2=Car("Wagnor")

s2.show()
```
