```python
# Else block run only when no exception occurs
try:
  num=int(input())


except ValueError:
  print("Invalid input")
else:
  print("You entered:",num)
```

```python
try:
  num=int(input())


except ValueError:
  print("Invalid input")
else:
  print("You entered:",num)
```

```python
  # Finally block always executes irrespective whether there is an error or not
try:
  num=int(input())


except ValueError:
  print("Invalid input")
finally:
  print("Program finished")
```
