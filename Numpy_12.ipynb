```python
import numpy as np
```

```python
# A bank account has the following transactions:
# transactions = np.array([
# 5000, -2000, 15000, -4500, 7000,
# -12000, 25000, -3000, 8000, -18000,
# 4500, 22000, -6000, 9000, -2500
# ])
# Positive values represent deposits and negative values represent withdrawals.



# Extract deposits.
# Extract withdrawals.
# Calculate total deposits and withdrawals.
# Calculate net transaction amount.
# Calculate absolute transaction values.
```

```python
transactions = np.array([
5000, -2000, 15000, -4500, 7000,
-12000, 25000, -3000, 8000, -18000,
4500, 22000, -6000, 9000, -2500
])
```

```python
# Extract deposits.

deposite=transactions[transactions >0]
deposite
```

```python
# Extract withdrawals.
withdraw=transactions[transactions <0]
withdraw
```

```python
# Calculate total deposits and withdrawals.
total_deposite = sum(deposite)
total_deposite
```

```python

total_withdraw = sum(withdraw)
total_withdraw
```

```python
# Calculate net transaction amount.
np.sum(transactions)
```

```python
# Calculate absolute transaction values.

abs_value=np.abs(transactions)
abs_value

```

```python
# Find transactions whose absolute value is above the average absolute transaction.
# Find the largest deposit.
# Find the largest withdrawal.
# Create labels "Credit" and "Debit"
# Find percentage of transactions whose absolute value is greater than ₹10,000.
# Reshape the data into 3 × 5.
# Calculate total transaction value for each row.
```

```python
# Find transactions whose absolute value is above the average absolute transaction.
avg=np.mean(abs_value)
transactions[abs_value > avg]
```

```python
# Find the largest deposit.

np.max(deposite)
```

```python
# Find the largest withdrawal.
np.max(withdraw)
```

```python
# Create labels "Credit" and "Debit"

label=np.where(transactions >0,"Credit","Debit")
label
```

```python
# Find percentage of transactions whose absolute value is greater than ₹10,000.

n=transactions[abs_value >10000]

percentage=(n.size/transactions.size)*100

percentage
```

```python
# Reshape the data into 3 × 5.
trans=transactions.reshape(3,5)
trans
```

```python
# Calculate total transaction value for each row.
value=np.sum(trans , axis=1)
value
```

```python
# A factory has four machines operating for six days:
# production = np.array([
# [120, 135, 128, 145, 150, 138],
# [110, 125, 118, 130, 142, 135],
# [160, 155, 170, 165, 180, 175],
# [95, 105, 100, 115, 108, 120]
# ])
# Calculate total production by machine.
# Calculate total production by day.
# Calculate average production of each machine.
# Find the machine with the highest average production.
# Find the day with the lowest production.

```

```python
production = np.array([
[120, 135, 128, 145, 150, 138],
[110, 125, 118, 130, 142, 135],
[160, 155, 170, 165, 180, 175],
[95, 105, 100, 115, 108, 120]
])
```

```python
# Calculate total production by machine.
total_production=np.sum(production , axis=1)
total_production
```

```python
# Calculate total production by day.
day_production=np.sum(production , axis=0)
day_production
```

```python
# Calculate average production of each machine

each_machine=np.mean(production , axis=1)
each_machine


```

```python
# Find the machine with the highest average production.

avg1=np.mean(production ,axis=1)
np.argmax(avg1)
```

```python
# Find the day with the lowest production.

mn=np.min(production,axis=0)
np.argmin(mn)
```

```python
# Calculate annual interest for every loan.
# Calculate total interest.
# Calculate average interest.
# Find the loan generating the highest interest.
# Find loans above ₹400,000 with interest rate above 9%.

# loan_amount = np.array([
# 100000, 250000, 500000, 750000,
# 120000, 350000, 900000, 450000
# ])

# interest_rate = np.array([
# 8.5, 9.2, 8.8, 10.1,
# 7.9, 9.5, 10.5, 8.7
# ])

```

```python
loan_amount = np.array([
100000, 250000, 500000, 750000,
120000, 350000, 900000, 450000
])
```

```python
interest_rate = np.array([
8.5, 9.2, 8.8, 10.1,
7.9, 9.5, 10.5, 8.7
])
```

```python
# Calculate annual interest for every loan.
annual=(loan_amount*interest_rate)/100
annual
```

```python
# Calculate total interest.
total=np.sum(annual)
total
```

```python
# Calculate average interest
avg=np.mean(annual)
avg
```

```python
# Find the loan generating the highest interest.
np.argmax(annual)
```

```python
# Find loans above ₹400,000 with interest rate above 9%.
loan_amount[np.where((loan_amount >400000) & (interest_rate >9))]
```

```python
# Categorize loans:
# Rate >= 10 → High
# Rate >= 9 → Medium
# Otherwise → Low

catg=np.where(interest_rate >=10,"High",np.where(interest_rate >=9,"Medium","Low"))
catg
```

```python
# Round interest to 2 decimal places.
np.round(interest_rate,2)
```

```python
# Calculate standard deviation of interest rates.
np.std(interest_rate)
```

```python
# Find median of loan amount
med = np.median(loan_amount)
med
```

```python
avg=np.mean(interest_rate)
avg
```

```python
# Find loans above median amount AND interest above average interest.

criteria = np.where((loan_amount > med) & (interest_rate >avg))
loan_amount[criteria]
```
