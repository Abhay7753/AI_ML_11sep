```python
inventory = np.array([
[12, 45, 8, 32, 20],
[5, 60, 15, 25, 10],
[30, 18, 6, 40, 22],
[9, 55, 12, 28, 7]
])
```

```python
# Calculate total inventory per store.

total = np.sum(inventory , axis=1)
total
```

```python
# Calculate total inventory per product.
total1 = np.sum(inventory , axis=0)
total1
```

```python
# Find the store with highest inventory.

store =np.argmax(total)
ans = store+1
ans
```

```python
# Find the product with lowest total inventory.

prd = np.argmin(total1)

ans1 = prd+1
ans1
```

```python
# Calculate the change between consecutive days.
# Calculate absolute price movement.
# Find the largest movement.
# Find average movement.
# Find movements above average.
# Calculate percentage change between consecutive days.
# Round percentage changes to 2 decimals.
# Find days where price increased by more than 2%.
# Find maximum, minimum and range.
# Reshape prices into 3 × 5.
# Find average price of each row
```

```python
# Calculate the change between consecutive days.
a = prices[1:]
a
```

```python
b= prices[:-1]
b
```

```python
change = a-b
change
```

```python
# Calculate absolute price movement.

absolute = np.abs(change)
absolute
```

```python
# Find the largest movement.
larg = np.max(change)
larg
```

```python
# Find average movement
avg = np.mean(change)
avg
```

```python
# Find movements above average.

avg_above = change[change >avg]
avg_above
```

```python
# Calculate percentage change between consecutive days.

percentage = np.round((change*100)/prices[:-1],2)
percentage
```

```python
# Round percentage changes to 2 decimals.

np.round(percentage ,2)
```

```python
# Find days where price increased by more than 2%.

incre = percentage[percentage > 2]
incre


```
