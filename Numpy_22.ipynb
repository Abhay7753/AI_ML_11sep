```python
def find_missing(arr,n):
  expected_sum=n*(n+1)/2
  total=0
  for i in arr:
    total+=i

  return expected_sum-total

print("Missing number ",find_missing(arr,n))
```

```python
def missing(arr,n):

  for i in range(1,n+1):
    if i not in arr:
      return i
print("Missing:",missing(arr,n))
```

```python
# Find Two Numbers Whose Sum Equals a Target
# Problem statement: Given an array and a target value,
# return the indices of two distinct elements whose sum equals the target.
# Assume exactly one valid pair exists.
# Input: arr = [2, 7, 11, 15], target = 9
# Output: [0,1]
```

```python
arr = [2,7,11,15]
target =9

seen ={}

def two_sum(arr,target):
  for i, num in enumerate(arr):
    required = target-num
    if required in seen:
      return[seen[required],i]

    seen[num] = i
  return []

print(two_sum(arr,9))
```

```python
# Problem statement: Two strings are anagrams
# if they contain the same characters with the same frequencies, regardless of order.
# Ignore spaces and case.
```

```python
# Input:
# s1: "Listen"
# s2: "Silent"
# output: "Yes"
```

```python
s1 ="Listen"
s2 ="Silent"

s1 ="".join(sorted(s1.lower()))
s2 ="".join(sorted(s2.lower()))

s1==s2
```

```python


def are_anagrams(s1, s2):
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()
    if len(s1) != len(s2):
      return False
    frequency = {}
    for char in s1:
       frequency[char] = frequency.get(char, 0) + 1
    for char in s2:
      if char not in frequency:
         return False
      frequency[char] -= 1
      if frequency[char] < 0:
        return False
    return True
print(are_anagrams("Listen", "Silent"))
print(are_anagrams("Hello", "World"))

```

```python
# Problem statement: Given a string, find the first character that appears exactly once.
# Return -1 if every character repeats.
# Input: "aabbcddee"
# Output: 'c'
# Input: "aabcddee"
# Output: 'b'
```

```python

def repeat(s):

  count ={}

  for char in s:
    count[char] = count.get(char,0)+1


  for i in count:
    if count[i]==1:
      return i
print("Output",repeat("aabcddee"))

```

```python

# Problem statement:
# Given two arrays sorted in ascending order,
# merge them into one sorted array without sorting the combined result afterward
# Input: [1, 4, 7, 9] and [2, 3, 8, 10]
# Output: [1, 2, 3, 4, 7, 8, 9, 10]
```

```python
num1=[1, 4, 7, 9]
num2= [2, 3, 8, 10]
final=[]
i=0
j=0

while i<len(num1) and j<len(num2):

  if num1[i] < num2[j]:
    final.append(num1[i])
    i+=1
  else:
    final.append(num2[j])
    j+=1


while i<len(num1):
  final.append(num1[i])
  i+=1
while j<len(num2):
  final.append(num2[j])
  j+=1

print("Final",final)
```
