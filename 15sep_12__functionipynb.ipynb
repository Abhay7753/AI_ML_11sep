#Positional argument

```python
def gr(name,age):
  print("Name is",name)
  print("Age is",age)
```

```python
gr("Abhay",20)
```

```python
gr("Aman",20)
```

#Keywoded argument


```python
def gr(name,age):
  print("Name is",name)
  print("Age is",age)
```

```python
gr(name="abhay",age=20)
```

```python
gr(age=20,name="Abhay")
```
