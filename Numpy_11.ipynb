```python
# An e-commerce company records the quantity sold and selling price of 12 products.
# quantity = np.array([2, 5, 3, 8, 4, 6, 10, 1, 7, 5, 9, 3])
# price = np.array([
# 1200, 800, 1500, 450, 2200, 900,
# 700, 1800, 650, 1300, 500, 2500
# ])
# Calculate revenue for every product.
# Calculate the total revenue.
# Find the product index generating the highest revenue.
# Extract the revenue of the last five products
```

```python
quantity = np.array([2, 5, 3, 8, 4, 6, 10, 1, 7, 5, 9, 3])
price = np.array([
1200, 800, 1500, 450, 2200, 900,
700, 1800, 650, 1300, 500, 2500
])
```

```python
# Calculate revenue for every product

revenue=quantity*price
revenue
```

```python
# Calculate the total revenue.
np.sum(revenue)
```

```python
# Find the product index generating the highest revenue.
np.argmax(revenue)
```

```python
# Extract the revenue of the last five products
revenue[-5:]
```

```python
# Find products whose revenue is above the average revenue.
# Apply a 10% discount only to products whose price is greater than ₹1,000.
# Round the discounted prices to two decimal places.
# Calculate the total quantity sold.
# Find the median product price.
# Identify products whose price is above the median and whose quantity is at least 5
```

```python
# Find products whose revenue is above the average revenue.
a=np.mean(revenue)
revenue[revenue > a]
```

```python
np.where(revenue > a, "Yes", "No")
```

```python
# Apply a 10% discount only to products whose price is greater than ₹1,000.
discount=np.where(price >1000 ,0.90*price,price)
discount
```

```python
# Round the discounted prices to two decimal places
np.round(discount,2)
```

```python
# Calculate the total quantity sold.
np.sum(quantity)
```

```python
# Find the median product price.
m=np.median(price)
m
```

```python
# Identify products whose price is above the median and whose quantity is at least 5.

price[(price >m)  & (quantity >=5)]
```

```python
# Create a forecast for the next 6 months using np.linspace() between:
# the minimum observed sales
# and 20% above the maximum observed sales.
# Round the forecast values to the nearest rupee.
# Combine the original sales and forecast into one 18-element array.
# Reshape the combined array into 3 × 6.
# Calculate the average sales for each row.
# Identify values that are greater than the overall median


```

```python
forecast = np.linspace(np.min(sales),0.20*np.max(sales),6)
forecast
```

```python
# Round the forecast values to the nearest rupee.
np.round(forecast,3)
```

```python
# Combine the original sales and forecast into one 18-element array.

s=np.append(sales , forecast)

s

```

```python
# Reshape the combined array into 3 × 6.
r=s.reshape(3,6)
r
```

```python
# Calculate the average sales for each row.
np.round(np.mean(r, axis=1),)
```

```python
# Identify values that are greater than the overall median
med=np.median(r)

r[r > med]
```

```python
# A factory collects temperatures from 20 sensor readings:
# temperature = np.array([
# 28.5, 31.2, 29.8, 35.6, 38.1,
# 40.2, 27.9, 30.5, 42.3, 39.8,
# 36.4, 33.2, 29.1, 41.7, 44.2,
# 31.8, 34.5, 37.6, 26.9, 43.1
# ])
# Calculate mean, median, minimum and maximum.
# Calculate variance and standard deviation.
# Find all temperatures more than one standard deviation above the mean.
# Calculate the absolute difference between every temperature and the mean.
# Find the five readings having the largest deviation from the mean.
# Round all temperatures to the nearest whole number.
```

```python
temperature = np.array([
28.5, 31.2, 29.8, 35.6, 38.1,
40.2, 27.9, 30.5, 42.3, 39.8,
36.4, 33.2, 29.1, 41.7, 44.2,
31.8, 34.5, 37.6, 26.9, 43.1
])
```

```python
# Calculate mean, median, minimum and maximum.
print(np.mean(temperature))

print(np.median(temperature))
print(np.min(temperature))
print(np.max(temperature))
```

```python
# Calculate variance and standard deviation.
print(np.std(temperature))
print(np.square(np.std(temperature)))
```

```python
# Find all temperatures more than one standard deviation above the mean.

temperature[temperature >(np.mean(temperature)+np.std(temperature)) ]
```

```python
# Calculate the absolute difference between every temperature and the mean.
deviation=np.abs(temperature -np.mean(temperature))
deviation
```

```python
# Find the five readings having the largest deviation from the mean.

largest=np.sort(deviation)

largest[-5:]

```

```python
# Round all temperatures to the nearest whole number.

np.round(temperature)
```
