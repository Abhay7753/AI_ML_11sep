```python
# [value_if_true if condition else value_if_false for item in iterable]
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
# even_odd = [...... "Even", "Odd"]

even_odd =[ "Even" if n%2==0 else "Odd" for n in numbers]

even_odd
```

```python
# We can use nested conditional expressions when there are three categories.
# Marks ≥ 80 → "Distinction"
# Marks ≥ 40 → "Pass"
# Otherwise → "Fail"
# marks = [95, 72, 38, 81, 45, 29]
marks = [95, 72, 38, 81, 45, 29]
```

```python
# Problem statement: Given employee salaries, calculate bonuses using these rules:
# Salary ≥ ₹80,000: 20% bonus
# Salary ≥ ₹50,000 but below ₹80,000: 10% bonus
# Salary below ₹50,000: 5% bonus
# Return a list containing the bonus amounts, not the updated salaries

salaries = [90000, 65000, 40000, 100000, 50000]

bonus =[ salary*0.20 if salary >=80000 else  salary*0.10 if salary >=50000
        else salary*0.05 for salary in salaries]
bonus

```

```python
# Given two lists, find elements that are present in the first list but not in the second list.
# List 1: [10, 20, 30, 40]
# List 2: [20, 40, 50]
# ans: [10, 30]

num1 = [10, 20, 30, 40]
num2 = [20, 40, 50]

num =[]
for a in num1:
  if a not in num2:
    num.append(a)
num
```

```python
# Given a list of integers, determine whether the list is sorted in ascending order.
# Do not use sort() or sorted().
# Input: [10, 20, 30, 40]
# Output: Sorted
# Input: [10, 30, 20, 40]
# Output: Not Sorted
```

```python

nums =[10,20,30,40]


for i in range(len(nums)-1):
  if (nums[i]<nums[i+1]):
    print("Not sorted")
    break

else:
  print("Sorted")
```
