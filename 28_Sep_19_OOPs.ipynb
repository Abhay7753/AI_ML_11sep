```python
class Student:
  def __init__(self,name,age,city):
    self.name=name
    self.age=age
    self.city=city
  def display_profile(self):
    print(self.name,self.age,self.city)
```

```python
s2=Student("Rajensh chauhan",25,"Gorakhpur")

s2.display_profile()
```

# Mini-case study

```python
class BankHolder:
  def __init__(self,holder,balance):
    self.holder=holder
    self.balance=balance
  def deposit(self,amount):
    self.balance+=amount

  def withdraw(self,amount):
    if self.balance>=amount:
     self.balance-=amount
    else:
      print("Insufficient balance")
  def show_balance(self):
    print(self.balance)

```

```python
bank_acc1=BankHolder("Abhay jeet ",10000)
```

```python
bank_acc1.holder
```

```python
bank_acc1.show_balance()
```

```python
bank_acc1.deposit(455858)
```

```python
bank_acc1.show_balance()
```

```python
bank_acc1.withdraw(1500)
```

```python
bank_acc1.show_balance()
```

```python

bank_acc1.withdraw(15005858488)
```
