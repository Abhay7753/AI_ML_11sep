# Problem 3: Remove Duplicate names

```python
# Problem statement
# take names separated by spaces.
# remove duplicate names while preserving the first occurence

# eg1: Rahul Amit Rahul Neha Amit
```

```python
names=input()
s=names.split(" ")# jaha pr space milega waha se splot kr dega word ko in sentence mae

print(s)
```

```python
unique_names=[]

for name in s:
  if name not in unique_names:
    unique_names.append(name)
print(unique_names)
```

#Problem 4: Character counter

```python
#  Take a string as input from the user

# Display  1.Number of vowels 2. number of consonant 3.number of digit 4.number of spaces


```

```python
words=input()

countv=0
countc=0
countd=0
counts=0

for ch in words:
  if ch in "aeiouAEIOU":
    countv+=1
  elif ch == " ":
    counts+=1
  elif ch.isdigit():
    countd+=1
  else:
    countc+=1
print("vowel",countv)
print("consonant",countc)
print("Digit",countd)
print("Space",counts)

```

```python
names=[]
prices=[]

for i in range(5):
  name=input()
  price=int(input())
  names.append(name)
  prices.append(price)
print(names)
print(prices)
```

```python
total_bill_amount=sum(prices)
mx=prices[0]
mx_prd=""
mn=prices[0]
mn_prd=""

for i in range(5):
  if prices[i]>mx:
    mx_prd=names[i]
    mx=prices[i]
  elif prices[i]<mn:
    mn_prd=names[i]
    mn=prices[i]
print(total_bill_amount)
print(mx,mx_prd)
print(mn,mn_prd)
```
