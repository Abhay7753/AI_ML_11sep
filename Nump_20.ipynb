```python
# Generate pairs whose sum is greater than 12
numbers1 = [1, 2, 3]
numbers2 = [10, 20]

pairs =[]

for n1 in numbers1:
  for n2 in numbers2:
    if n1+n2 >12:
     pairs.append((n1,n2))
print(pairs)
```

```python
[(n1,n2)  for n1 in numbers1 for n2 in numbers2 if n1+n2 >12]
```

```python
# Find pairs of numbers from 1 to 5 where both numbers are different and their product is even.


ans =[]

for n1 in range(1,6):
  for n2 in range(1,6):
    if (n1!=n2) and (n1*n2)%2==0:
      ans.append((n1,n2))

ans
```

```python
# Extract only passing marks from every row
# marks >= 75
marks = [
[75, 80, 90],
[60, 85, 70],
[95, 88, 92]
]

passed = []

for mark in marks:
  for m in mark:
    if m>=75:
      passed.append(m)
print(passed)
```
