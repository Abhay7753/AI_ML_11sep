```python
# The coordinates of eight delivery locations are:
# points = np.array([
# [3, 4],
# [5, 12],
# [8, 15],
# [7, 24],
# [9, 40],
# [12, 35],
# [15, 20],
# [6, 8]
# ])
# Each row contains:
# [x, y]
# Calculate distance from the origin using:
# distance = √(x² + y²)
```

```python
# Extract x and y coordinates.
# Calculate distance for every location.
# Round distances to 2 decimals.
# Find nearest and farthest location.
# Find average distance.
# Find locations above average distance.
# Categorize:
# if >= 30 → Long
# elseif >= 15 → Medium
# Otherwise → Short
# Find percentage of Long-distance deliveries.
# Flatten the coordinate array.
# Find the size of the flattened array.
```

```python
import numpy as np
points = np.array([
[3, 4],
[5, 12],
[8, 15],
[7, 24],
[9, 40],
[12, 35],
[15, 20],
[6, 8]
])
```

```python
# Extract x and y coordinates.
x = points[: ,0]
x
```

```python
x1 = np.square(x)
x1
```

```python
y = points[: ,1]
y
```

```python
y1 =np.square(y)
y1
```

```python
# Calculate distance for every location.

dist = np.sqrt(x1 + y1)
dist
```

```python
# Round distances to 2 decimals.
np.round(dist ,2)
```

```python
# Find nearest and farthest location.
near=np.min(dist)
near
```

```python
far = np.max(dist)
far
```

```python
# Find average distance.
avg = np.mean(dist)
avg
```

```python
# Find locations above average distance.

loct = np.where( dist > avg)
points[loct]
```

```python
# Categorize:
# if >= 30 → Long
# elseif >= 15 → Medium
# Otherwise → Short

category = np.where(dist >=30 ,"Long",np.where( dist >=15 ,"Medium","Short"))
category
```

```python
# Find percentage of Long-distance deliveries.
long = category[np.where( category == "Long")]
long

```

```python
c = len(long)
d= len(dist)

```

```python
percentage = (c/d)*100
percentage
```

```python
# Flatten the coordinate array.

flat = points.flatten()
flat
```

```python
# Find the size of the flattened array.

flat.size
```
