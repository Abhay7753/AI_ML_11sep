```python
# Consider the following employee data:
# employees = np.array([
# [101, 42000, 3, 450000],
# [102, 65000, 6, 720000],
# [103, 38000, 2, 300000],
# [104, 90000, 8, 950000],
# [105, 55000, 5, 610000],
# [106, 72000, 7, 830000]
# ])
# Columns represent:
# Employee_ID
# Salary
# Experience
# Annual_Sales
```

```python
# Extract salary, experience and annual sales.
# Calculate sales-to-salary ratio.
# Find the employee with the highest annual sales.
# Find employees having at least 5 years experience AND sales above ₹600,000.
# Calculate bonus:
# Sales >= ₹800,000 → 10%
# Sales >= ₹600,000 → 7%
# Otherwise → 3%
```

```python
import numpy as np
employees = np.array([
[101, 42000, 3, 450000],
[102, 65000, 6, 720000],
[103, 38000, 2, 300000],
[104, 90000, 8, 950000],
[105, 55000, 5, 610000],
[106, 72000, 7, 830000]
])
```

```python
# Extract salary, experience and annual sales.
sales=employees[: ,3]
salary=employees[:,1]
experience=employees[:,2]
```

```python
salary
```

```python
sales
```

```python
experience
```

```python
# Calculate sales-to-salary ratio
np.round(sales/salary)

```

```python
# Find the employee with the highest annual sales.
employees[np.argmax(sales)]
```

```python
# Find employees having at least 5 years experience AND sales above ₹600,000.
employees[(employees[: ,3]) >=5 & (employees[: ,3] > 600000)]
```

```python
# Calculate bonus:
# Sales >= ₹800,000 → 10%
# Sales >= ₹600,000 → 7%
# Otherwise → 3%

bonus=np.where(sales >=800000,sales*0.1,np.where(sales>=600000 ,sales*0.07,sales*0.03))

bonus

```

```python
# calculate total bonus amount.
# Find average salary of employees receiving a 10% bonus.
# Find median salary.
# Find employees whose salary is above median salary but experience is less than 6 years.

# Round sales-to-salary ratio to 2 decimal places.
```

```python
# Consider the following 6 × 6 grayscale image:
# image = np.array([
# [10, 25, 40, 55, 70, 85],
# [20, 35, 50, 65, 80, 95],
# [30, 45, 60, 75, 90, 105],
# [40, 55, 70, 85, 100, 115],
# [50, 65, 80, 95, 110, 125],
# [60, 75, 90, 105, 120, 135]
# ])
# Find ndim, shape and size.
# Extract the central 4 × 4 region.
# Increase the central region by 20%.
# Cap all pixel values above 100 at 100.
# Find mean, minimum and maximum pixel values.
```

```python
image = np.array([
[10, 25, 40, 55, 70, 85],
[20, 35, 50, 65, 80, 95],
[30, 45, 60, 75, 90, 105],
[40, 55, 70, 85, 100, 115],
[50, 65, 80, 95, 110, 125],
[60, 75, 90, 105, 120, 135]
])
```

```python
# Find ndim, shape and size.
image.ndim

```

```python
image.shape
```

```python
image.size
```

```python
# Extract the central 4 × 4 region.

central=image[1:5,1:5]
central
```

```python
# Increase the central region by 20%
image[1:5,1:5]=image[1:5,1:5]*1.2
image
```

```python
# Cap all pixel values above 100 at 100.
image=np.where(image >100,100,image)
image
```

```python
# Find mean, minimum and maximum pixel values.

np.mean(image)
```

```python
np.min(image)
```

```python
np.max(image)
```

```python
# Find pixels above the 75th percentile.
# Flatten the image.
# Find pixels below the overall mean.
# Reshape the flattened array back into 6 × 6.
```

```python
# Find pixels above the 75th percentile.
p75=np.percentile(image,75)
p75


print(image[image >p75])
```

```python
# Flatten the image.

img=image.flatten()
img
```

```python
# Find pixels below the overall mean.
mn=np.mean(image)

image[image <mn]
```

```python
# Reshape the flattened array back into 6 × 6.

img.reshape(6,6)
```
