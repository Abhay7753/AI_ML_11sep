#Broadcasting

```python
import numpy as np

marks =np.array([70,80,90])

updated_marks =marks+5

print(updated_marks)
```

```python
marks = np.array([70,80,90])

updated_marks = np.zeros(len(marks),dtype =int)

for i in range(len(marks)):
  updated_marks[i] = marks[i]+5
print(updated_marks)
```

```python
# Categorize each student as:
# - High attendance: 85% or above
# - Moderate attendance: 70% to below 85%
# - Low attendance: below 70%
# - Unknown: attendance is missing
```

```python
catg = np.where(np.isnan(attendance),"Missing",(np.where(attendance >=85,"High attendence",np.where(attendance >=70,"Moderate attendance","Low"))))

catg
```

```python
conditions = [
np.isnan(attendance),
attendance >= 85,
attendance >= 70
]
choices = [
"Unknown",
"High",
"Moderate"
]
attendance_category = np.select(
conditions,
choices,
default="Low"
)
print(attendance_category)
```

```python
# Categorize students into three study bands:
# less than 2 hours, 2–4 hours inclusive, and more than 4 hours

study_band = np.where(
study_hours < 2,
"Low",
np.where(study_hours <= 4, "Medium", "High")
)
print(study_band)
```
