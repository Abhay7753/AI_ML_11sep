```python
# Given the age of person ,check if they are eligible to vote
# if they are at atleast 20 they are eligible else not eligible

age=int(input())
if age>=20:
  print("eligible")
else :
  print("Not eligible")
```

```python
# Take the password from the user . if it matches "hello@123" then print "Welcome"
password=input()

if password=="hello@123":
  print("welcome")
else:
  print("invalid")
```
