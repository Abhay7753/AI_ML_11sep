```python
# Print all the int from 1 to 5
for i in range(1,6):
  print(i)
```

```python
i=1
while i<6:
  print(i)
  i=i+1
```

```python
# Write while loop to print HI 7 times on the screen
i=1
while i<8:
  print("hi")
  i=i+1
```

```python
#Print all the even numbers between 0 to 10 (both inclusive) using a while loop
i=0
while i<=10:
  if i%2==0:
    print(i)
  i+=1
```

```python
# Print all the odd numbers between 0 to 10
i=1
while i<=10:
  if i%2!=0:
    print(i)
  i=i+1;
```

```python
# Print all the odd numbers between 0 to 10
i=1
while i<=10:

    print(i+2)
    i+=2
```

```python
# Print the sum of all number 1 to 10 both inclusive
i=1;
sum=0;
while i<=10:
  sum+=i
  i+=1
print(sum)
```

```python
# Print sum of 1 to 10 number inclusive both by for loop
sum=0
for i in range (11):
  sum+=i
print(sum)
```

```python
# Print the multiplication tavle of 5 till its first 10 multiples:
i=1
while i<=10:
  print("5 *",i,"=",5*i)
  i+=1
```
