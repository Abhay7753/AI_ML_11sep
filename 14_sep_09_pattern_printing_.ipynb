#Recap session


```python
for i in range (1,10):
  if i==4:
    break
  print(i)
```

```python
for i in range (1,10):
  if i==4:
    continue
  print(i)
```

# Nested loop


```python
for i in range(5):
  for j in range(30):
    print("i=",i,"j=",j)
```

```python
# pattern printing
for i in range (4):
  for j in range(5):
    print("*",end=" ")
  print()
```

```python
for i in range(5):
  for j in range(5):
    print(i+1,end=" ")
  print()
```

```python
for i in range(5):
  for j in range(5):
    print(j+1,end=" ")
  print()
```

```python
for i in range (3):
  for j in range(4):
    print((j+1)*10,end=" ")
  print()
```

```python
for i in range (5):
  for j in range (i+1):
    print("*",end=" ")
  print()
```

```python
for i in range (5):
  for j in range(i+1):
    print(i+1,end=" ")
  print()
```

```python
for i in range (5):
  for j in range (i+1):
    print(j+1,end=" ")
  print()
```

```python
n=1
for i in range (1,5):
  for j in range (i):
    print(n,end=" ")
    n+=1
  print()
```
