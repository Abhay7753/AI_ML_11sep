# Higher order Functions (HOF)

```python
# Hof are functions that:
# 1 - accepts another functions as an argument
# 2- returns another functions
```

```python
def add(a,b):
  return a+b

def mul(a,b):
  return a*b


def operation(f,a,b):
  return f(a,b)

```

```python
ans = operation(add,3,5)
ans
```

```python
# Given a list of integers find the largest element without using Pythonis built- in max() function.


# Input = [12,45,7,89,34]
# output =89

Input = [12,45,7,89,34]

mx = 0
for i in Input:
  if i > mx:
    mx =i
print("Largest numbers:",mx)
```

```python
# Find second largest number in the  list without sorting


numbers =[12,45,7,89,34]

mx1 =float('-inf')

mx2=float('-inf')


for num in numbers:
  if num > mx1:
    mx2=mx1
    mx1=num
  elif num > mx2:
      mx2=num

print(mx2)
```

```python
# Given a list of integers create  a new list containing only unique elements:

# Maintain the original order of elements.


x = [10,20,10,30,20,40]
unique =[]

for num in x:
  if num not in unique:
    unique.append(num)
print("Unique elements:",unique)
```

```python
# Move all Zeros to the end

# given a list of integer , move all zeros to the end of the list while maintaining the relative order of the no_zero element


# Input : [0,5,0,3,8,0,2]
```

```python
x = [0,5,0,3,8,0,2]
```

```python
arr =[]
zeros = []
for i in x:
  if i == 0:
    zeros.append(i)
  else:
    arr.append(i)
print(arr+zeros)
```

```python
# Given a list of numbers calculate the average nd print all element that are greater than the average
# without numpy

y =[10,20,30,40,50]

total =sum(y)
avg =total/len(y)

```

```python
result =[]
for num in y:
  if num > avg:
    result.append(num)

print("Number:",result)
print("average",avg)
```

```python
number =[10,-5,0,8,-3,12,-7]
pos =[]
neg =[]

for n in number:
  if n != 0:
    if n<0:

       neg.append(n)
    else:
      pos.append(n)

print("Positive",pos)
print("Negative",neg)
```
