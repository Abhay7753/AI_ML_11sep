```python
# Password validator
# Problem statement
#  A password is  valid if it contains
# minimum 8 characters ,atleast one uppercase,atleast one lowercase,atleast one digit
# Dispaly strong password or weak password


```

```python
password=input()

countupper=0
countlower=0
countdigit=0

for i in password:
  if i.isupper():
    countupper+=1
  elif i.islower():
    countlower+=1
  elif i.isdigit():
    countdigit+=1

if countupper>0 and countlower>0 and countdigit>0 and len(password)>7:
  print("strong password")
else:
  print("Weak password ")
```

 # Most frequent character

```python
# Problem statement
# Find the character that appears the maximum number of times in a string ,ignore spaces
```

```python
s="Python is most amazing programming language for Ml today"

freq={}

for c in s:
  if c!=" ":
   if c not in freq:
    freq[c]=1
   else:
    freq[c]+=1
print(freq)
```

```python
mx=0
for key in freq:
  if freq[key]>mx:
    mx=freq[key]
print(mx)
```

```python
for k,v in freq.items():
  print(k,v)
```

```python
mx_v=0
for k,v in freq.items():

  if v>mx_v:
    mx_v=v
    mx_key=k
print(mx_key,mx_v)
```

```python
# __init__()
# __init__()----> is built-in method which is always executed when the class
# gets initiated (object is being)

```

```python
class Person:
  def __init__(self,name,city):
    self.name=name
    self.city=city
```

```python
p1=Person("Raj","Mumbai")
print(p1.name,p1.city)
```

```python
p2=Person("Abhayjeet","Gorakhpur")

print(p2.name,p2.city)
```
