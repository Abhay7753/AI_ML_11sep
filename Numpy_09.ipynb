```python
# Consider :
# employees = np.array([    [101, 35000, 2],
#                       [102, 48000, 4],
#                           [103, 62000, 6],
#                           [104, 40000, 3],
#                           [105, 75000, 8],
#                           [106, 55000, 5] ])

# Column represent:
# Employee Id | Salary | Experience

# Find:
# Employees earning more than ₹50,000.
# Employees with more than 4 years of experience.
# Employees earning more than ₹50,000 AND having more than 4 years of experience.
# Average salary of employees satisfying condition 3.
# Employee with the highest salary.
# Employees earning below the median salary.
```

```python
employees = np.array([    [101, 35000, 2],
                      [102, 48000, 4],
                          [103, 62000, 6],
                          [104, 40000, 3],
                          [105, 75000, 8],
                          [106, 55000, 5] ])
```

```python
# Employees earning more than ₹50,000.

employees[np.where(employees[: , 1] > 50000)]
```

```python
# Employees with more than 4 years of experience.

employees[employees[: ,2] > 4]
```

```python
# Employees earning more than ₹50,000 AND having more than 4 years of experience.

employees[(employees[:,1] > 50000) & (employees[:,0]>4)]
```

```python
# Average salary of employees satisfying condition 3.
arr=employees[(employees[:,1] > 50000) & (employees[:,0]>4)]
np.mean(arr[:,1])
```

```python
# Employee with the highest salary.
np.max(employees[:,1])
```

```python
# Employees earning below the median salary.
employees[np.median(employees[:,1])<employees[:,1]]
```
