#Continue


```python
# continue skip the current iteration
for i in range(1,10):
  if i==5:
    continue
  print(i)
```

```python
for i in range (1,10):
  print(i)
  if i==5:
    continue
```

```python
# Print all the even numbers between 1 to 10 (both inclusive)--continue statement
for i in range (1,10):
  if i%2!=0:
    continue
  print(i)
```
