```python
students = [
{
"roll":101,
  "name":"Rahul",
"age":20,
"course":"Python",
"marks":85
},
{
"roll":102,
"name":"Neha",
"age":19,
"course":"SQL",
"marks":92
},
{
"roll":103,
"name":"Amit",
"age":22,
"course":"Python",
"marks":74
},
{
"roll":104,
"name":"Priya",
"age":21,
"course":"AI",
"marks":96
},
{
"roll":105,
"name":"Rohan",
"age":20,
"course":"Python",
"marks":67
}
]
```

```python
# Print only the names of all students

for student in students:
  print(student["name"])
```

```python
# Calculate the total marks obtained by all student:
total =0
for student in students:
    print(student["marks"])
    total+=student["marks"]
print(total)
```

```python
# Find average Marks of the class

n=len(students)
total =0
for student in students:
    print(student["marks"])
    total+=student["marks"]
print(total/n)


```

```python
# Count passed mark  ---> passing mark is 40;
count=0
for student in students:
  if student["marks"]>=90:
    count+=1

print(count)
```

```python
# Write a function to print grade of every students
#  Rules a-->90+ b-->75+ c--60+ d--baki


def grade(num):
  if num>90:
    return "A+"
  elif num>75:
    return "B+"
  elif num>60:
    return "C+"
  else:
    return "d"
```

```python
for student in students:
  print(student["marks"],grade(student["marks"]))
```

```python
# create a function to search student by their roll no

def search(roll_no):
  for student in students:
    if student["roll"]==roll_no:
      return student["name"]


```

```python
# search student by name

n=input()

for student in students:
  if student["name"].lower()==n.lower():
    print(student)
```

```python
# Write a function to print the topper of the class

def find_topper(students):
  mx_marks=0
  topper={}
  for student in students:
    if student["marks"] > mx_marks:
      mx_marks=student["marks"]
      topper=student
  return topper


```

```python
ans=find_topper(students)
print(ans)
```

```python
# Count student in each course

countp=0
countai=0
countsql=0
for student in students:
  if student["course"]=="Python":
    countp+=1
  elif student["course"]=="SQL":
    countsql+=1
  else:
    countai+=1

print(countsql,countp,countsql)
```

```python
courses={}
for student in students:
  if student["course"] not in courses:
    courses[student["course"]]=1
  else:
    courses[student["course"]]+=1
print(courses)
```

```python
# Remove  dulicate courses

courses=[]

for student in students:
  courses.append(student["course"])
print(courses)
s=set(courses)
print(s)
```

```python
# Create student class (oop) in the above format
class Student:
  def __init__(self,roll,name,age,course,marks):
    self.roll=roll
    self.name=name
    self.age=age
    self.course=course
    self.marks=marks
  def show_student(self):
    print(self.roll,self.name,self.age,self.course,self.marks)
```

```python
s1=Student(106,"Mj",30,"Dance",100)
```

```python
s1.show_student()
```
