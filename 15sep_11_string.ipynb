```python
marks=[65,75,85]
total=0
for m in marks:
  if m>=70:
    total+=m
print(total)
```

```python
names=["Amit","Abhayjeet","AMan"]
longest=names[0]
for i in names:
  if len(names)>len(longest):
    longest=names
print(longest)
```

```python
data =[10,20,30]
for i in range(len(data)):
  data[i]=data[i]+10
print(max(data))
print(min(data))
```

```python
numbers =[5,10,15,20]
total=0
for i in range(len(numbers)):
  if i%2==0:
    total+=numbers[i]
print(total)
```
