```python
# Take a number from the user and check whether it is even or odd

number=int(input())
if number%2==0:
  print("even")
else:
  print("odd")
```

```python
#Take two numbers and print the larger number .if both are equal ,print "Both numbers are equal"
a=int(input())
b=int(input())
if a>b:
  print(a)
elif b>a:
  print(b)
else:
  print("Both number are equal")
```

```python
# calculate bmi (body mass index)
#  less than 18.5 underweight
# 18.25 to 24.9 normal weight
# 25.0to 29.9 overweight
# 30.0or above obese
age=float(input())

if age<18.5:
  print("underweight")
elif age<=24.9:
  print("normal weight")
elif age<=29.9:
  print("overweight")
else:
  print("obese")

```

```python
# Take integer and Check divisibility
a=int(input())

if (a%3==0) and(a%5==0):
  print("Divisible by both 3 and 5")
elif (a%3==0) and(a%5!=0):
  print("divisible by only 3")
elif (a%5==0) and(a%3!=0):
  print("divisible by 5 only  ")
else:
  print("not divisible by 5 or 3")
```
